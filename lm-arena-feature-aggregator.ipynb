````python
# ═══════════════════════════════════════════════════════════════════════
# Cell 3 — DATASET_FEATURES GROUP: Load from "Dataset _features/"
# ═══════════════════════════════════════════════════════════════════════
#
# IMPORTANT: The folder is "Dataset _features" (with a SPACE before _)
#
# These functions extract features from the `conv_metadata` column
# (nested dict already in the dataset), NOT from raw response text!
#
# Notebooks found:
#   F_Unified Dataset Feature.ipynb  → extract_dataset_format_features(df)  [MASTER]
#   F_Header Feature.ipynb           → extract_header_features(meta)
#   F_Bold Text Feature.ipynb        → extract_bold_features(meta)
#   F_List Feature.ipynb             → extract_list_features(meta)
#   F_Token Feature.ipynb            → extract_token_features(meta)
#   F_Conversation Dynamics Features.ipynb → extract_conversation_dynamics(meta)
#
# ═════════════════════════════════════════════════════════���═════════════

import os, sys, ast, json, re, math, types, importlib.util

REPO_DIR = "LM_Arena_140K_TR"

# ────────────────────────────────────────────────────────────
# Helper 1: load_notebook  — execute .ipynb cells, return namespace
# ────────────────────────────────────────────────────────────
def load_notebook(path):
    """Execute all code cells in a .ipynb and return the resulting namespace."""
    with open(path, 'r', encoding='utf-8') as f:
        nb = json.load(f)
    ns = {}
    # Pre-inject common imports so notebook code can use them
    exec("import re, math, ast, warnings\nfrom typing import Any, Dict, List, Optional, Tuple, Set", ns)
    for cell in nb.get('cells', []):
        if cell['cell_type'] == 'code':
            source = ''.join(cell['source'])
            if source.strip():
                try:
                    exec(source, ns)
                except Exception as e:
                    print(f"  ⚠ Skipped a cell in {os.path.basename(path)}: {e}")
    return ns

# ────────────────────────────────────────────────────────────
# Helper 2: load_module  — import a .py file by path
# ────────────────────────────────────────────────────────────
def load_module(name, path):
    """Import a .py file as a module given its file path."""
    spec = importlib.util.spec_from_file_location(name, path)
    mod = importlib.util.module_from_spec(spec)
    spec.loader.exec_module(mod)
    return mod

# ────────────────────────────────────────────────────────────
# Load Dataset _features notebooks
# ────────────────────────────────────────────────────────────
dataset_features_dir = os.path.join(REPO_DIR, "Dataset _features")

# Unified (master function)
nb_unified = load_notebook(os.path.join(dataset_features_dir, "F_Unified Dataset Feature.ipynb"))
dataset_parse_metadata = nb_unified['parse_metadata']
dataset_get_value = nb_unified['get_value']
dataset_extract_dataset_format_features = nb_unified['extract_dataset_format_features']

# Individual feature notebooks
nb_header = load_notebook(os.path.join(dataset_features_dir, "F_Header Feature.ipynb"))
dataset_extract_header_features = nb_header['extract_header_features']

nb_bold = load_notebook(os.path.join(dataset_features_dir, "F_Bold Text Feature.ipynb"))
dataset_extract_bold_features = nb_bold['extract_bold_features']

nb_list = load_notebook(os.path.join(dataset_features_dir, "F_List Feature.ipynb"))
dataset_extract_list_features = nb_list['extract_list_features']

nb_token = load_notebook(os.path.join(dataset_features_dir, "F_Token Feature.ipynb"))
dataset_extract_token_features = nb_token['extract_token_features']

nb_conv = load_notebook(os.path.join(dataset_features_dir, "F_Conversation Dynamics Features.ipynb"))
dataset_extract_conversation_dynamics = nb_conv['extract_conversation_dynamics']

print("✅ DATASET_FEATURES GROUP — 6 functions + 2 helpers loaded")
print()
print("  Master (applies to whole df via conv_metadata):")
print("    • dataset_extract_dataset_format_features(df) → df with 22+ columns")
print()
print("  Individual extractors (work on parsed metadata dict):")
print("    • dataset_extract_header_features(meta)")
print("    • dataset_extract_bold_features(meta)")
print("    • dataset_extract_list_features(meta)")
print("    • dataset_extract_token_features(meta)")
print("    • dataset_extract_conversation_dynamics(meta)")
print()
print("  Helpers (available for all subsequent cells):")
print("    • load_notebook(path)   — execute .ipynb, return namespace")
print("    • load_module(name, path) — import .py file as module")
````

````python
# ═══════════════════════════════════════════════════════════════════════
# Cell 9f — SAVE THE ENRICHED DATASET
# ═══════════════════════════════════════════════════════════════════════

print("📦 SAVING...")

# Convert list columns to string for CSV compatibility
if 'response_a_code_languages' in df.columns:
    df['response_a_code_languages'] = df['response_a_code_languages'].apply(str)
    df['response_b_code_languages'] = df['response_b_code_languages'].apply(str)

# Save CSV
output_csv = 'lm_arena_140k_all_features.csv'
df.to_csv(output_csv, index=False)
print(f"✅ CSV saved: {output_csv}")
print(f"   Shape: {df.shape}")
print(f"   Size:  {os.path.getsize(output_csv) / (1024**3):.2f} GB")

# Save Parquet (smaller, preserves dtypes)
output_parquet = 'lm_arena_140k_all_features.parquet'
df.to_parquet(output_parquet, index=False)
print(f"✅ Parquet saved: {output_parquet}")
print(f"   Size:  {os.path.getsize(output_parquet) / (1024**3):.2f} GB")

# Summary
feature_cols = [c for c in df.columns if any(c.startswith(p) for p in [
    'response_a_', 'response_b_', 'dataset_a_', 'dataset_b_',
    'total_', 'header_', 'bold_', 'emphasis_', 'ordered_', 'unordered_',
    'assistant_', 'user_', 'context_', 'turns', 'is_multi_turn', 'turn_density',
    'prompt_'
])]
print(f"\n📊 Total feature columns: {len(feature_cols)}")
print(f"📊 Total columns: {df.shape[1]}")
print(f"📊 Total rows: {df.shape[0]:,}")
````
